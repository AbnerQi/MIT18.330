### 练习 1.1.3

⌨ 在例 1.1.2 中，我们使用了 $\frac{22}{7}$ 作为 $\pi$ 的有理数近似。下面给出的两个有理数都是比 $\frac{22}{7}$ 更好的 $\pi$ 的近似。

对于下面的每一个近似值，计算：

1. 绝对准确度；
2. 相对准确度；
3. 准确数字的位数，并将结果向下取整为整数。

**(a)**
$
\frac{355}{113}
$

**(b)**
$
\frac{103638}{32989}
$

````python
# (a)

x = 355 / 113

print(f"computed value: {x}")
print(f"standard value: {pi}")
print(f"absolute accuracy: {abs(pi - x)}")
rel_acc = abs(pi - x) / pi
print(f"relative accuracy: {rel_acc: .4e}")
print(f"accurate digits: {math.floor(-log10(rel_acc))}")
````

````python
# (b)

x = 103638 / 32989

print(f"computed value: {x}")
print(f"standard value: {pi}")
print(f"absolute accuracy: {abs(pi - x)}")
rel_acc = abs(pi - x) / pi
print(f"relative accuracy: {rel_acc: .4e}")
print(f"accurate digits: {math.floor(-log10(rel_acc))}")
````

### 练习 1.1.5（Python / NumPy 版）

⌨ NumPy 提供函数 `np.nextafter(x, direction)`，用于返回从浮点数 `x` 出发、朝 `direction` 方向的下一个可表示浮点数。

本题使用双精度类型 `np.float64`。其中：

- `np.finfo(np.float64).max` 表示最大的有限双精度浮点数；
- `np.inf` 表示正无穷；
- `-np.inf` 表示负无穷。

**(a)** 最大的有限双精度浮点数之后，沿正无穷方向的下一个浮点数是什么？

**(b)** 从负无穷出发，沿正无穷方向的下一个浮点数是什么？

使用 NumPy 编程验证：

```python
import numpy as np
```

````python
import numpy as np

# (a) 
with np.errstate(over="ignore"):
    print(np.nextafter(np.finfo(np.float64).max, np.inf))
````

````python
# (b) 

print(np.nextafter(-np.inf, np.inf))
print(np.finfo(float64).min)
````
